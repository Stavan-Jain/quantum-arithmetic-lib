```python
import numpy as np
import pandas as pd
from fractions import Fraction
from math import floor, gcd, log

from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit.circuit.library import QFT, UnitaryGate
from qiskit.transpiler import CouplingMap, generate_preset_pass_manager
from qiskit.visualization import plot_histogram

from qiskit_ibm_runtime import QiskitRuntimeService
from qiskit_ibm_runtime import SamplerV2 as Sampler
```

```python
def swap01():
    U = QuantumCircuit(3)
    U.swap(0, 1)

    U = U.to_gate()
    return U
```

```python
swap01_gate = swap01()

circ = QuantumCircuit(3)
circ.compose(swap01_gate, inplace = True)
circ.decompose(reps = 2).draw(output = "mpl", fold = -1)
```

```python
def controlled_swap01():
    U = QuantumCircuit(3)
    U.swap(0, 1)

    U = U.to_gate()
    c_U = U.control()

    return c_U
```

```python
controlled_swap01_gate = controlled_swap01()
circ = QuantumCircuit(4)
# use append which accepts Instruction/ControlledGate without type issues
circ.append(controlled_swap01_gate, qargs=[0, 1, 2, 3])
circ.decompose(reps = 1).draw(output = "mpl",fold = -1)
```
